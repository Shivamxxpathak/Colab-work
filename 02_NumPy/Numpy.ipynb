```python
#to check how many dimension array is having
#we use ndim attribute
import numpy as np
a = np.array(42)
b = np.array([1,2,3,4])
c = np.array([[1,2,3],[4,5,6]])
d = np.array([[[1,2,3],[4,5,6]],[[1,2,3],[4,5,6]]])
print(a.ndim)
print(b.ndim)
print(c.ndim)
print(d.ndim)
```

```python
#higher dimensional arrays--we define the no of dimensons by using the
#attribute ndmin
import numpy as np
arr = np.array([1,2,3,4,5],ndmin=5)
print(arr)
print('number of dimensions:',arr.ndim)
```

Resulting Shape: (1, 1, 2, 2, 3)

Dimensions 0 and 1 were added because of ndmin=5 (sizes are 1).

Dimensions 2, 3, and 4 come from the shape of the input data itself (2 blocks, 2 rows, 3 columns).

```python
# ufuncs --universal functions
# vectorization --process of iterative statements into vector-based operations

x = [1,2,3,4]
y = [5,6,7,8]
z = np.add(x,y)
print(z)

#create your own ufunctions
def myadd(x,y):
  return x+y
myadd = np.frompyfunc(myadd,2,1)
print(myadd([1,2,3,4],[5,6,7,8]))

# ufuncs oprations add(), subtract(),multiply(),power(),mod(),remainder(),divmod(),exp(),log(),sqrt(),minimum(),maximum(),sum(),cumsum(),prod(),cumprod(),diff()
# add(),subtract(),divide(),multiply()
import numpy as np
arr1 = np.array([10,11,12,13,14,15])
arr2 = np.array([20,21,22,23,24,25])
newarr = np.add(arr1,arr2)
newarr1 = np.subtract(arr1,arr2)
newarr2 = np.divide(arr1,arr2)
newarr3 = np.multiply(arr1,arr2)
print(newarr)
print(newarr1)
print(newarr2)
print(newarr3)
```

```python
# join numpy arrays
# putting the content of 2 or more arrays in single array thru
# concatenate function concatenate()
import numpy as np
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])
arr = np.concatenate((arr1,arr2))
print(arr)
```

```python
# class task--join two 2-D arrays along axis
import numpy as np
arr1 = np.array([[1,3,5],[2,4,6]])
arr2 = np.array([[7,9,11],[8,10,12]])
arr = np.concatenate((arr1,arr2),axis=1)
print(arr)
```

```python
# join arrays using stack functions --it is same as concatenate the only difference is that stacking is done along new axis
#Type of Stacks-- hstack(),vstack(),dstack()
import numpy as np
arr1 = np.array([1,2,3])
arr2 = np.array([4,5,6])
arr = np.stack((arr1,arr2),axis=1)
arr = np.hstack((arr1,arr2)) #creates 1-d array
arr = np.vstack((arr1,arr2)) #creates 2-d array
arr = np.dstack((arr1,arr2)) #creates 3-d array
print(arr)
```
