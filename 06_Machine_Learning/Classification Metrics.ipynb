```python
# classification models predicts categorical values the metrices used are
#1)Accuracy--proportion of correct predictions made by model out of all predictions
#Accuracy = Number of correct predictions/total number of predictions
# in case of imbalanced datasets where a dataset with 90%class A
# 10% of class B,accuracy will predict only for class and will fail to
# identify class B
#2)Precision Precision = TP/TP+FP where TP = True positives,FP = False Positives
# it measures how many of the positive predictions made by model are actually correct
# it is useful when cost of the FP is high such as in medical diagnosiswhere predicting
# a disease when it is not present can have serious consequences
#3)Recall/Sensitivity:TP/TP+FN , where FN =  False Negatives measures how many of the
# actual positive cases were coreectly identified by the model. it is
# important when a missing positive case(False Negative FN) is more costly than
# FP(False Positive) recall is a key metrics in medical diagnosis
#4) F1 Score is the harmonic mean of precision and recall
# if F1 score is high means the model performs well on both metrics(precision & recall)
# lower reacll and higher precision gives great accuracy but then it misses large no. of
# instances , more the F1 score higher will be the performance
# F1 score = 2 X precisionxrecall/Precision+recall range of F1 score is (0,1)
# AUC-ROC Curve --Evaluates The model over a spectrum of thresholds where
# precision,recall & F1Score provides insights about a model over only a single threshold
# ROC Curve is graphical representation of TPR true postive rates vs the FPR false
# postive rates at different classification thresholds
# TPR = TP/TP+FN TP-true positive(correctly predicted positive cases)
# FN-false negative(actual positive cases incorrectly predicted as negative)
#TPR measures -- out of all actual positve cases how many did the model correctly identify
#TNR True negative rate: TN/TN+FP TN-true negative(correctly predicted negative cases)
#FP-false positive(actual negative cases incorrectly predicted as positive )
#TNR measure how many actual negative instances were correctly identified by the model or we can say--out of all actual negative cases how many did the model correctly identify as negative
```
