```python
# Matplotlib is a low level graph library that serves as a visualiztion utility
# Matplotlib is open sourcesource code of matplotlib is @http://github.com/matplotlib/matplotlib
# pip install matplotlib
# import matplotlib
# import matplotlib.pyplot as plt

import matplotlib.pyplot as plt
import numpy as np
xpoints = np.array([0,6])
ypoints = np.array([0,250])
plt.plot(xpoints,ypoints)
plt.show()
```

```python
# Multiple plots
# we have to draw from position(1,3) to position(2,8)
#

import numpy as np
import matplotlib.pyplot as plt
xpoint = np.array([1,4,5,7])
ypoint = np.array([3,9,6,14])
plt.plot(xpoint,ypoint)
plt.show()
```

```python
# to plot only marker we can use'o'
import numpy as np
import matplotlib.pyplot as plt
xpoint = np.array([1,4,5,7])
ypoint = np.array([3,9,6,14])
plt.plot(xpoint,ypoint,'d' , c = 'hotpink')
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt
xpoint = np.array([1,4,5,7])
ypoint = np.array([3,9,6,14])
plt.plot(xpoint,ypoint, ls = 'dashdot',c = 'hotpink',lw = 11)
plt.grid()
plt.show()
```

```python
# / labels
import numpy as np
import matplotlib.pyplot as plt
xpoint = np.array([1,4,5,7])
ypoint = np.array([3,9,6,14])
plt.plot(xpoint,ypoint, ls = 'dashdot',c = 'hotpink',lw = 1)
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.grid(True)
plt.show()
```
